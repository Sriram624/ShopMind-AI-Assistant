## 1. Data Ingestion (Amazon Reviews 2023)

```
dataset_id = 'McAuley-Lab/Amazon-Reviews-2023'
subset = 'raw_review_Electronics'
raw_stream = load_dataset(dataset_id, subset, split='train', streaming=True, trust_remote_code=True)
sample_list = [entry for i, entry in enumerate(raw_stream) if i < 5000]
df = pd.DataFrame(sample_list)
```

## 2. Low-Level Preprocessing

```
porter = PorterStemmer()
def clean_text(text):
    return ' '.join([porter.stem(w) for w in str(text).lower().split()])
df['cleaned_text'] = df['text'].apply(clean_text)
```
